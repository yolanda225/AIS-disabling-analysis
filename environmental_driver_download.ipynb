```python
import copernicusmarine
from config import PERIODS, CMEMS_USERNAME, CMEMS_PASSWORD

def download_sst(years):
    start = []
    end = []
    for year in years:
        start1, end1 = (f"{year}-01-01", f"{year}-12-31")
        start.append(start1)
        end.append(end1)  
    for start, end in zip(start, end):
        print("Starting download for period:", start, "to", end)
        copernicusmarine.subset(
            dataset_id="cmems_obs-mob_glo_phy-sal_my_multi-oi_P7D-c",
            variables=["analysed_sst"],
            start_datetime=start,
            end_datetime=end,
            output_filename=f"sst_{start}_{end}.nc",
            output_directory=str("data/raw/driver_data"),
            username=CMEMS_USERNAME,
            password=CMEMS_PASSWORD,
        )

#download_sst(["2017", "2018", "2019"])

# # have to do in chunks (device memory full)
# copernicusmarine.subset(
#   dataset_id="cmems_obs-mob_glo_phy-sal_my_multi-oi_P7D-c",
#   variables=["sst"],
#   minimum_longitude=-180,
#   maximum_longitude=179.75,
#   minimum_latitude=-80,
#   maximum_latitude=90,
#   start_datetime="2017-01-01T00:00:00",
#   end_datetime="2019-12-31T00:00:00",
#   minimum_depth=0.4940253794193268,
#   maximum_depth=0.4940253794193268,
#   output_filename=f"sst_{2017}_{2019}.nc",
#     output_directory=str("data/raw/driver_data/cmems"),
#     username=CMEMS_USERNAME,
#     password=CMEMS_PASSWORD,
# )

#chl
import copernicusmarine

copernicusmarine.subset(
  dataset_id="cmems_mod_glo_bgc_my_0.25deg_P1M-m",
  variables=["chl"],
  minimum_longitude=-180,
  maximum_longitude=179.75,
  minimum_latitude=-80,
  maximum_latitude=90,
  start_datetime="2023-01-01T00:00:00",
  end_datetime="2025-12-31T00:00:00",
  minimum_depth=0.5057600140571594,
  maximum_depth=0.5057600140571594,
  output_filename=f"chl_{2023}_{2025}.nc",
    output_directory=str("data/raw/driver_data/cmems"),
    username=CMEMS_USERNAME,
    password=CMEMS_PASSWORD,
)


```
